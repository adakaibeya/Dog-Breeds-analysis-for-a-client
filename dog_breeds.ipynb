A client, the owner of Happy paws pet service, is planning to open a dog adoption and pet recommendation service. Primary customers are families with young children, disabled individuals in need of emotional support animals, and individuals looking for service dogs. 
Families with young children are looking for dogs that are good with children, low barking tendencies of 1, manageable in size?(below average in height and weight), males and females inclusive, low shedding tendencies, and playful.
Disabled individuals are looking for dogs that are playful and protective, low barking tendencies, manageable in size(below average in height and weight).
Service dogs that have above average max_life_expectancy, above average height, preferably male dogs, protective score of 4-5, trainability score of 4-5, and energy score of 4-5.

```python
dog_breeds.describe()
```

```python
male_average_dog_height=dog_breeds['max_height_male'].mean()
female_average_dog_height=dog_breeds['max_height_female'].mean()
print(male_average_dog_height)
print(female_average_dog_height)
average_dog_height = (male_average_dog_height + female_average_dog_height) / 2
average_dog_height
```

#For families with young children


```python
suitable_dog_breeds_family = dog_breeds[(dog_breeds['max_height_female'] < average_dog_height) & (dog_breeds['max_height_male'] < average_dog_height) & (dog_breeds['good_with_children'].isin([4,5])) & (dog_breeds['barking'].isin([1,2])) & (dog_breeds['shedding'].isin([2,3])) & (dog_breeds['playfulness'].isin([4,5]))]
suitable_dog_breeds_family
```

```python
df1=suitable_dog_breeds_family[['Name', 'max_height_male', 'max_height_female', 'good_with_children', 'barking', 'shedding', 'playfulness']].reset_index(drop=True)
df1
```

#for disabled individuals 

```python
suitable_dog_breeds_disabled = dog_breeds[(dog_breeds['max_height_female'] < average_dog_height) & (dog_breeds['max_height_male'] < average_dog_height) & (dog_breeds['protectiveness'].isin([4,5])) & (dog_breeds['barking'].isin([1,2])) & (dog_breeds['playfulness'].isin([2,3]))]
df2=suitable_dog_breeds_disabled[['Name', 'max_height_male', 'max_height_female', 'protectiveness', 'barking', 'playfulness']].reset_index(drop=True)
df2
```

#dog breeds for people in service

```python
avg_max_life_expectancy = dog_breeds['max_life_expectancy'].mean()
avg_max_life_expectancy
```

```python
suitable_dog_breeds_service = dog_breeds[(dog_breeds['max_height_male'] > average_dog_height) & (dog_breeds['max_life_expectancy'] >= avg_max_life_expectancy) & (dog_breeds['protectiveness'].isin([4,5])) & (dog_breeds['energy'].isin([4,5])) & (dog_breeds['trainability'].isin([4,5]))]
df3=suitable_dog_breeds_service[['Name', 'max_height_male', 'max_life_expectancy', 'protectiveness', 'energy', 'trainability']].reset_index(drop=True)
df3
```
